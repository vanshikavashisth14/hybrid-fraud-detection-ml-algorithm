```python
# ===========================
# FraudFusion Prototype
# Isolation Forest + CatBoost
# ===========================

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import IsolationForest
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from catboost import CatBoostClassifier

# -----------------------------
# Load Dataset
# -----------------------------
df = pd.read_excel("/content/synthetic_upi_fraud_dataset (1).xlsx")

# -----------------------------
# Separate encoder for each column
# -----------------------------
encoders = {}

categorical_cols = ["Merchant","Device","Location","Payment_Mode"]

for col in categorical_cols:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col].astype(str))
    encoders[col] = le

# -----------------------------
# Time Feature
# -----------------------------
df["Transaction_Time"] = pd.to_datetime(df["Transaction_Time"])
df["Hour"] = df["Transaction_Time"].dt.hour
df.drop("Transaction_Time", axis=1, inplace=True)

# -----------------------------
# Features
# -----------------------------
X = df.drop("Fraud", axis=1)
y = df["Fraud"]

# -----------------------------
# Isolation Forest
# -----------------------------
iso = IsolationForest(
    contamination=0.08,
    random_state=42
)

X["Anomaly_Score"] = iso.fit_predict(X)
X["Anomaly_Score"] = X["Anomaly_Score"].replace({1:0,-1:1})

# -----------------------------
# Split
# -----------------------------
X_train,X_test,y_train,y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# -----------------------------
# CatBoost
# -----------------------------
model = CatBoostClassifier(
    iterations=100,
    learning_rate=0.1,
    depth=6,
    verbose=False
)

model.fit(X_train,y_train)

# -----------------------------
# Prediction
# -----------------------------
pred=model.predict(X_test)

print("\nAccuracy :",accuracy_score(y_test,pred))
print("\nClassification Report\n")
print(classification_report(y_test,pred))

print("\nConfusion Matrix\n")
print(confusion_matrix(y_test,pred))

# ====================================================
# Test New Transaction
# ====================================================

merchant="Amazon"
device="Android"
location="Delhi"
payment="UPI"

# Safe Encoding
def safe_encode(col,value):
    le=encoders[col]
    if value in le.classes_:
        return le.transform([value])[0]
    else:
        return -1

new=pd.DataFrame([{
    "Amount":50000,
    "Merchant":safe_encode("Merchant",merchant),
    "Device":safe_encode("Device",device),
    "Location":safe_encode("Location",location),
    "Payment_Mode":safe_encode("Payment_Mode",payment),
    "Hour":2
}])

new["Anomaly_Score"]=iso.predict(new)
new["Anomaly_Score"]=new["Anomaly_Score"].replace({1:0,-1:1})

result=model.predict(new)

print("\nPrediction :","Fraud" if result[0]==1 else "Genuine")
```

```python
# ==========================================================
# FraudFusion++
# Hybrid UPI Fraud Detection Framework
# Isolation Forest + CatBoost + ARDE
# ==========================================================

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import IsolationForest
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    roc_curve,
    auc
)

from catboost import CatBoostClassifier
import matplotlib.pyplot as plt

# ==========================================================
# LOAD DATASET
# ==========================================================

df = pd.read_excel("/content/synthetic_upi_fraud_dataset (1).xlsx")

# ==========================================================
# LABEL ENCODING
# ==========================================================

encoders = {}
categorical_cols = ["Merchant","Device","Location","Payment_Mode"]

for col in categorical_cols:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col].astype(str))
    encoders[col] = le

# ==========================================================
# TIME FEATURE
# ==========================================================

df["Transaction_Time"] = pd.to_datetime(df["Transaction_Time"])
df["Hour"] = df["Transaction_Time"].dt.hour
df.drop("Transaction_Time", axis=1, inplace=True)

# ==========================================================
# FEATURES
# ==========================================================

X = df.drop("Fraud", axis=1)
y = df["Fraud"]

# ==========================================================
# ISOLATION FOREST
# ==========================================================

iso = IsolationForest(
    contamination=0.08,
    random_state=42
)

X["Anomaly_Score"] = iso.fit_predict(X)
X["Anomaly_Score"] = X["Anomaly_Score"].replace({1:0,-1:1})

# ==========================================================
# TRAIN TEST SPLIT
# ==========================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# ==========================================================
# CATBOOST MODEL
# ==========================================================

model = CatBoostClassifier(
    iterations=100,
    learning_rate=0.1,
    depth=6,
    verbose=False
)

model.fit(X_train, y_train)

# ==========================================================
# TEST RESULTS
# ==========================================================

pred = model.predict(X_test)
prob = model.predict_proba(X_test)[:,1]

print("\nAccuracy :",round(accuracy_score(y_test,pred),3))

print("\nClassification Report\n")
print(classification_report(y_test,pred))

print("\nConfusion Matrix\n")
print(confusion_matrix(y_test,pred))

# ==========================================================
# ROC CURVE
# ==========================================================

fpr,tpr,_=roc_curve(y_test,prob)
roc_auc=auc(fpr,tpr)

plt.figure(figsize=(6,5))
plt.plot(fpr,tpr,label="AUC = %.3f"%roc_auc)
plt.plot([0,1],[0,1],'--')
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.legend()
plt.show()

# ==========================================================
# SAFE ENCODER
# ==========================================================

def safe_encode(col,value):
    le=encoders[col]
    if value in le.classes_:
        return le.transform([value])[0]
    return -1

# ==========================================================
# ARDE
# Adaptive Risk Decision Engine
# ==========================================================

def arde(probability, amount, anomaly, hour):

    reasons=[]

    if amount>30000:
        reasons.append("High transaction amount")

    if anomaly==1:
        reasons.append("Anomalous transaction detected")

    if hour<6 or hour>23:
        reasons.append("Late-night transaction")

    if probability>=0.85:
        risk="HIGH"
        decision="FRAUD"

    elif probability>=0.65 and anomaly==1:
        risk="MEDIUM"
        decision="FRAUD"

    elif probability>=0.50:
        risk="LOW"
        decision="REVIEW"

    else:
        risk="SAFE"
        decision="GENUINE"

    if len(reasons)==0:
        reasons.append("Transaction pattern appears normal")

    return risk,decision,reasons

# ==========================================================
# NEW TRANSACTION
# ==========================================================

merchant="Amazon"
device="Android"
location="Delhi"
payment="UPI"

new=pd.DataFrame([{

    "Amount":50000,

    "Merchant":safe_encode("Merchant",merchant),

    "Device":safe_encode("Device",device),

    "Location":safe_encode("Location",location),

    "Payment_Mode":safe_encode("Payment_Mode",payment),

    "Hour":2

}])

new["Anomaly_Score"]=iso.predict(new)
new["Anomaly_Score"]=new["Anomaly_Score"].replace({1:0,-1:1})

prediction=model.predict(new)[0]

probability=model.predict_proba(new)[0][1]

risk,decision,reasons=arde(

    probability,

    new["Amount"].iloc[0],

    new["Anomaly_Score"].iloc[0],

    new["Hour"].iloc[0]

)

# ==========================================================
# FINAL OUTPUT
# ==========================================================

print("\n==============================")
print("FraudFusion++ Prediction")
print("==============================")

print("Prediction          :", "Fraud" if prediction==1 else "Genuine")
print("Fraud Probability   : {:.2f}%".format(probability*100))
print("Risk Level          :",risk)
print("Decision            :",decision)
print("Isolation Forest    :",
      "Anomaly" if new["Anomaly_Score"].iloc[0]==1 else "Normal")

print("\nReason(s):")

for r in reasons:
    print("-",r)
```
